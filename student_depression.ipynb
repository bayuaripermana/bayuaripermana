```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
df = pd.read_csv("/content/Student_Depression_Dataset.csv", encoding='utf-8-sig')
df.head()
```

```python
print('Jumlah baris & kolom: ',df.shape)
print()
df.info()
```

```python
df.describe()
```

```python
df.describe(include='object')
```

```python
df.raw = df.copy()
df.columns = (df.columns
              .str.strip()
              .str.lower()
              .str.replace(' ', '_'))

df = df.rename(columns={
    'cost_without_study': 'monthly_cost_without_study',
    'friend_circle': 'has_friend_circle'
})

print(df.columns.tolist())
```

```python
print('Missing values per kolom')
print(df.isnull().sum())
print()
print('Jumlah baris duplikat', df.duplicated().sum())
```

```python
before = len(df)
df = df.drop_duplicates().reset_index(drop=True)
print(f'Baris dihapus: {before - len(df)}')

# add primary key
df.insert(0, 'student_id', range(1, len(df) + 1))
df.head()
```

```python
cat_cols = df.select_dtypes(include="object").columns

for col in cat_cols:
  df[col] = df[col].str.strip()

for col in cat_cols:
  print(f'---{col}---')
  print(df[col].value_counts(), '\n')
```

```python
# 6a. Rentang nilai tiap kolom numerik
print('Age              :', df['age'].min(), '-', df['age'].max())
print('CGPA             :', df['cgpa'].min(), '-', df['cgpa'].max(), '(skala 4.0)')
print('Current semester :', df['current_semester'].min(), '-', df['current_semester'].max())
print('Monthly cost     :', df['monthly_cost_without_study'].min(), '-', df['monthly_cost_without_study'].max())
```

```python
# 6b. Nilai mencurigakan
print('Semester = 0     :', (df['current_semester'] == 0).sum(), 'baris')
print('Biaya = 0        :', (df['monthly_cost_without_study'] == 0).sum(), 'baris')
print('Semester >= 13   :', (df['current_semester'] >= 13).sum(), 'baris')
print('Usia < 18        :', (df['age'] < 18).sum(), 'baris')
```

```python
# 6c. Cek logika antar kolom: status menikah vs status hubungan
pd.crosstab(df['marital_status'], df['relationship_status'])
```

```python
# Buat flag untuk baris yang perlu perhatian (tidak dihapus)
df['flag_zero_cost'] = df['monthly_cost_without_study'] == 0
df['flag_marital_conflict'] = (df['marital_status'] == 'Married') & (df['relationship_status'] == 'Single')

print('Flag zero cost        :', df['flag_zero_cost'].sum())
print('Flag marital conflict :', df['flag_marital_conflict'].sum())
```

```python
def iqr_outliers(series):
    q1, q3 = series.quantile([0.25, 0.75])
    iqr = q3 - q1
    low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    return series[(series < low) | (series > high)], low, high

for col in ['age', 'cgpa', 'monthly_cost_without_study']:
    out, low, high = iqr_outliers(df[col])
    print(f'{col:25s} batas wajar: {low:.2f} - {high:.2f} | outlier: {len(out)}')
```

```python
# 8a. Depression -> binary (memudahkan SUM/AVG di BI: rata-rata = % depresi)
df['depression_flag'] = df['depression'].map({'Yes': 1, 'No': 0})

# 8b. Kelompok usia
df['age_group'] = pd.cut(df['age'],
                         bins=[16, 19, 22, 25, 30],
                         labels=['17-19', '20-22', '23-25', '26-28'])

# 8c. Kategori CGPA (skala 4.0)
df['cgpa_category'] = pd.cut(df['cgpa'],
                             bins=[0, 2.99, 3.49, 3.79, 4.0],
                             labels=['Low (<3.0)', 'Average (3.0-3.49)',
                                     'Good (3.5-3.79)', 'Excellent (3.8-4.0)'])

# 8d. Tahun studi dari semester
def semester_to_year(s):
    if s == 0:  return 'New Student'
    if s <= 2:  return 'Year 1'
    if s <= 4:  return 'Year 2'
    if s <= 6:  return 'Year 3'
    if s <= 8:  return 'Year 4'
    return 'Year 5+'
df['study_year'] = df['current_semester'].apply(semester_to_year)

# 8e. Kategori pengeluaran (berdasarkan kuartil agar seimbang)
df['cost_category'] = pd.qcut(df['monthly_cost_without_study'], q=4,
                              labels=['Low', 'Medium', 'High', 'Very High'])

# 8f. Kolom Yes/No -> ringkas
df['has_friend_circle'] = df['has_friend_circle'].replace({'Yes': 'Has Friends', 'No': 'No Friends'})
assert df.isnull().sum().sum() == 0, "Ada missing value, cek kolom: " + str(df.columns[df.isnull().any()].tolist())
assert df['student_id'].is_unique, "student_id tidak unik"

print("✅ Semua validasi lolos")
df[['age', 'age_group', 'cgpa', 'cgpa_category', 'current_semester',
    'study_year', 'monthly_cost_without_study', 'cost_category', 'depression_flag']].head(10)
```

```python
# Ubah ke tipe category dengan urutan logis
df['health_condition'] = pd.Categorical(df['health_condition'],
        categories=['Good', 'Normal', 'Abnormal', 'Critical'], ordered=True)

df['family_background'] = pd.Categorical(df['family_background'],
        categories=['Lower Middle class', 'Middle class', 'Rich class'], ordered=True)

df['study_year'] = pd.Categorical(df['study_year'],
        categories=['New Student', 'Year 1', 'Year 2', 'Year 3', 'Year 4', 'Year 5+'], ordered=True)

# Kolom boolean -> integer (0/1) agar kompatibel dengan semua BI tools
for col in ['flag_zero_cost', 'flag_marital_conflict']:
    df[col] = df[col].astype(int)

# Rapikan urutan kolom
df = df[['student_id', 'gender', 'age', 'age_group', 'family_background',
         'relationship_status', 'marital_status', 'living_style', 'has_friend_circle',
         'health_condition', 'home_division', 'current_semester', 'study_year',
         'cgpa', 'cgpa_category', 'monthly_cost_without_study', 'cost_category',
         'depression', 'depression_flag', 'flag_zero_cost', 'flag_marital_conflict']]
df.info()
```

```python
# Validasi akhir
print('Shape akhir      :', df.shape)
print('Missing values   :', df.isnull().sum().sum())
print('Duplikat         :', df.duplicated().sum())
print('student_id unik  :', df['student_id'].is_unique)
print()
print('Distribusi Depression:')
print(df['depression'].value_counts(normalize=True).round(3) * 100)
```

```python
# Quick sanity check sebelum ke BI: tingkat depresi per segmen
print(df.groupby('health_condition', observed=True)['depression_flag'].mean().mul(100).round(1))
print()
print(df.groupby('study_year', observed=True)['depression_flag'].mean().mul(100).round(1))
```
