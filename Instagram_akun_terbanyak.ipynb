```python
import pandas as pd
from google.colab import files

uploaded = files.upload()
nama_file = list(uploaded.keys())[0]

df_raw = pd.read_csv(nama_file)
df = df_raw.copy()
```

```python
print(df.shape)
df.info()
print(df.isna().sum())
print((df.isna().mean() * 100).round(1))
print(df.duplicated().sum())
print(df['date'].nunique())
print(df['username'].duplicated().sum())
```

```python
df['date'] = pd.to_datetime(df['date'])
print(df['date'].dtype)
print(df['date'].min(), df['date'].max())
```

```python
nol = df[df['followers']==0]
print(nol[['username', 'name', 'type', 'rank']])
```

```python
df['is_followers_nol'] = df['followers'] == 0
```

```python
print(df[df['country'].isna()][['username', 'type']].head())
df['country'] = df['country'].fillna('Global')
df['countries'] = df['countries'].fillna('Global')
```

```python
if 'also_in' in df.columns:
    df['punya_kategori_kedua'] = df['also_in'].notna()
    df = df.drop(columns=['also_in'])
else:
    print("Kolom 'also_in' tidak ditemukan di DataFrame!")
```

```python
negara = df[['username', 'countries']].copy()
negara['negara'] = negara['countries'].str.split(';')
negara = negara.explode('negara')[['username', 'negara']].reset_index(drop=True)
```

```python
print(df['followers'].sum())
print(df.merge(negara, on='username')['followers'].sum())
```

```python
df = df.drop(columns=['countries'])
```

```python
assert df['username'].is_unique
assert df['followers'].ge(0).all()
print(df.shape, negara.shape)
print(df.dtypes)

df.to_csv('instagram_akun_clean.csv', index=False, encoding='utf-8-sig')
negara.to_csv('instagram_akun_negara.csv', index=False, encoding='utf-8-sig')
files.download('instagram_akun_clean.csv')
files.download('instagram_akun_negara.csv')
```
